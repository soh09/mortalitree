```python
import modal

vol = modal.Volume.from_name("mot")
entries = list(vol.iterdir("TreeBoxes_v0.13/images"))
print(f"{len(entries)} files in TreeBoxes_v0.13/images")
```

```python
import re
from pathlib import Path

tile_names = set()
for e in entries:
    name = Path(e.path).name
    tile_names.add(re.sub(r"_image_tile_\d+\.jpg$", "", name))

print(f"{len(tile_names)} unique tiles")
for t in sorted(tile_names):
    print(t)
```

```python
unsup = df[df['source'] == 'Weinstein et al. 2018 unsupervised']
filenames = unsup['filename'].unique()
sites = set(n.split('_')[1] for n in filenames)
years = set(n.split('_')[0] for n in filenames)
print(f"unique filenames: {len(filenames)}")
print(f"years: {sorted(years)}")
print(f"sites: {sorted(sites)}")
```

```python
for src in ['Weecology_University_Florida', 'NEON_benchmark', 'Radogoshi et al. 2021']:
    fns = df[df['source'] == src]['filename'].unique()
    sample = list(fns)[:5]
    print(f"\n{src}: {len(fns)} unique files")
    for s in sample:
        print(f"  {s}")
```
